# dataset of professors

```python
prof = pd.read_csv("professors.csv")
prof
```

```python
print (prof.columns.tolist())
display(prof.isna().sum())
```

```python
prof = prof[["id" , "name" , "university" , "cited_by_all" , "cited_by_since_2020" , "h_index_all" , "h_index_since_2020" , "i10_index_all" , "i10_index_since_2020"]]
prof
```

```python
prof.info()
```

```python
prof.describe()
```

```python
prof = prof[prof["cited_by_all"] != "Not Found"]
prof
```

<pre style= "color: purple;">the rows with "Not Found" data, deleted

566 from 49941 of the professors where deleted(1.1%)</pre>

```python
prof.info()
```

```python
prof["cited_by_all"] = pd.to_numeric(prof["cited_by_all"], errors="coerce")
prof["cited_by_since_2020"] = pd.to_numeric(prof["cited_by_since_2020"], errors="coerce")
prof["h_index_all"] = pd.to_numeric(prof["h_index_all"], errors="coerce")
prof["h_index_since_2020"] = pd.to_numeric(prof["h_index_since_2020"], errors="coerce")
prof["i10_index_all"] = pd.to_numeric(prof["i10_index_all"], errors="coerce")
prof["i10_index_since_2020"] = pd.to_numeric(prof["i10_index_since_2020"], errors="coerce")
prof.info()
```

<p style="color:purple">numeric datas changed into int</p>

<p style= "color: purple;">dataset of professors now is clean and ready to be used!</p>

# analyzes

<p style="background-color: lightpink">سوال:  کدام ۱۰ دانشگاه بیشترین مجموع استنادات را دارند؟</p>

```python
top_uni = prof.groupby("university").agg(n_professors=("h_index_all" , "count") , 
                                        total_citations=("cited_by_all" , "sum")).reset_index()
```

```python
top_uni = top_uni.merge(uni[["Rank" , "University" , "Town"]] , left_on="university" , right_on="University" , how="left")
```

```python
top_uni = top_uni.sort_values("total_citations" , ascending=False).reset_index(drop=True)
top10 = top_uni.head(10)
```

```python
display(top10[["university" , "Town" , "Rank" , "n_professors" , "total_citations"]])
```

<pre style="background-color: lightblue">تحلیل: دانشگاه علوم پزشکی تهران با ۵۱۱۴۰۰۴ استناد ٬ رتبه اول پژوهشی را دارد و اختلاف آن با دانشگاه دوم ۲۰۵۷۶۸۹ میباشد

    از ۱۰ دانشگاه برتر ٬ ۶ دانشگاه در تهران قرار دارد

 رنکینگ دانشگاه ها لزوما بر پایه ی پژوهشی نیست چون تاپ ۱۰ ما با ۱۰ دانشگاه با رنک برتر٬ یکسان نیست
</pre>

<p style="background-color: lightpink">سوال: اگر بجای حجم٬ کیفیت را ملاک بگذاریم رتبه بندی چه فرقی میکند؟</p>

```python
quality = prof.groupby("university").agg(n_professors=("h_index_all" , "count") , 
                                         mean_h=("h_index_all" , "mean") , 
                                         median_h=("h_index_all" , "median") , 
                                         max_h=("h_index_all" , "max") , 
                                         total_citation=("cited_by_all" , "sum")).reset_index()
```

فقط دانشگاه های با حداقل ۵۰ استاد٬ برای اینکه میانگین قابل اعتماد باشد

```python
quality_big = quality[quality["n_professors"] >= 50].copy()
```

```python
quality_big = quality_big.sort_values("mean_h" , ascending=False).reset_index(drop=True)
top10_q = quality_big.head(10)
```

```python
top10_q
```

```python
plt.barh(top10_q["university"] , top10_q["mean_h"] , color="violet")
plt.xlabel("mean_h_index")
plt.ylabel("University")
plt.title("top 10 universities by mean h_index (min 50 professors)")
```

<pre style="background-color: lightblue">تحلیل: رتبه بندی بر اساس مجموع استنادات با رتبه بندی بر اساس میانگین
h_index
یکسان نیست. دانشگاه علوم پزشکی تهران در رتبه حجمی اول است ولی در رتبه کیفی پنجم است٬ که نشان میدهد جایگاه آن بیشتر مدیون تعداد استاد است تا میانگین کیفیت. در مقابل پژوهشگاه پلیمر و پتروشیمی ایران با وجود تعداد استاد کمتر٬ میانگین 
h_index
بالاتری دارد</pre>

```python
quality["n_professors"].max()
```

```python
print(quality.loc[quality["n_professors"].idxmax() , ["university" , "n_professors"]])
```

<p style="background-color: yellow">بیشترین تعداد اساتید مربوط به دانشگاه تهران است</p>

<p style="background-color:lightpink">سوال: رابطه کمیت و کیفیت در دانشگاه ها چگونه است؟</p>

```python
plt.scatter(quality["n_professors"] , quality["mean_h"] , alpha = 0.6)
plt.xlabel("number of professors (quantity)")
plt.ylabel("mean h_index (quality)")
plt.title("quantity VS quality per university")
```

<pre>رابطه تعداد استاد و میانگین 
h_index
خطی نیست.</pre>

<p style="background-color: lightpink">سوال: پژوهش کشور چقدر متمرکز است؟</p>

```python
uni_cited = prof.groupby("university")["cited_by_all"].sum().sort_values(ascending=False)
total_all = uni_cited.sum()
top5_sum = uni_cited.head(5).sum()
share5 = top5_sum / total_all * 100
```

```python
print(f"کل استنادات کشور : {total_all : ,.0f}")
```

```python
print(f"مجموع ۵ دانشگاه اول : {top5_sum : ,.0f}")
```

```python
print(f"سهم ۵ دانشگاه اول : {share5 : ,.0f} درصد")
```

```python
sizes = list(uni_cited.head(5).values) + [total_all - top5_sum]
labels = list(uni_cited.head(5).index) + ["Others"]
plt.pie(sizes , labels=labels , autopct="%1.1f%%" , startangle=90)
plt.title("Share of Top 5 Universities VS Others")
```

```python
top5_unis = uni_cited.head(5).index
n_prof_top5 = prof[prof["university"].isin(top5_unis)].shape[0]
n_prof_total = len(prof)
```

```python
print(f"سهم ۵ دانشگاه اول از تعداد استادها : {n_prof_top5 / n_prof_total * 100 : .1f} درصد")
print(f"سهم ۵ دانشگاه اول از استنادات : {share5 : .1f} درصد")
```

<p style="background-color: lightblue"> جواب: پنج دانشگاه اول کشور با در اختیار داشتن ۲۶.۲ درصد از استادها٬ ۳۳.۶ درصد از کل استنادات را به خود اختصاص داده اند. این اختلاف نشان میدهد تمرکز پژوهش علاوه بر تعداد استاد٬ ناشی از بهره وری بالاتر این دانشگاه هاست</p>

<p style="background-color: lightpink">سوال: عملکرد ۵ سال اخیر با کل دوران چه فرقی دارد؟</p>

باید ببینیم هر دانشگاه چقدر از استنادش مال ۵ سال اخیر است. هرچه این سهم بیشتر باشد یعنی جوان تر و فعال تر است

```python
national_share = prof["cited_by_since_2020"].sum() / prof["cited_by_all"].sum() *100
print(f"سهم ۵ سال اخیر از کل استنادات کشور : {national_share : .1f} درصد")
```

سهم ۵ سال اخیر به تفکیک دانشگاه

```python
recent = prof.groupby("university").agg(
    n_professors=("h_index_all" , "count") , 
    total_all=("cited_by_all" , "sum") , 
    total_recent=("cited_by_since_2020" , "sum") , 
    mean_h_all=("h_index_all" , "mean") , 
    mean_h_recent=("h_index_since_2020" , "mean")
).reset_index()

recent["recent_share"] = recent["total_recent"] / recent["total_all"] * 100

recent_big = recent[recent["n_professors"] >= 100].copy()
recent_big = recent_big.sort_values("recent_share" , ascending=False).reset_index(drop=True)

display(recent_big[["university" , "n_professors" , "mean_h_all" , "mean_h_recent" , "recent_share"]].head(10))
```

<p style="background-color: orange ; display:inline">نمودار فعال ترین دانشگاه ها در ۵ سال اخیر</p>

```python
data = recent_big.head(10).sort_values("recent_share")
plt.barh(data["university"], data["recent_share"], color="lightgreen")
plt.xlabel("share of citation since 2020(%)")
plt.title("Most Active Universities In Last 5 Years")
```
